# HR Policies Q&A

```python
import pandas as pd

# Load the hr_policies_qa dataset from Hugging Face Hub
df = pd.read_json("hf://datasets/strova-ai/hr-policies-qa-dataset/dataset.jsonl", lines=True)
```

```python
df.head()
```

```python
for index, row in df.iterrows():
    print(f"row: {row[0][2]['content']}")
    print("-" * 50)
```

```python
# extract the question and answer pairs from the dataset and save them in a txt file
with open("qa_pairs.txt", "w") as f:
    for index, row in df.iterrows():
        question = row[0][1]['content']
        answer = row[0][2]['content']
        f.write(f"Question: {question}\n")
        f.write(f"Answer: {answer}\n")
        f.write("-" * 50 + "\n")
```
